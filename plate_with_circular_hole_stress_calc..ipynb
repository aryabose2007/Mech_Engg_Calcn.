```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
w = 0.3 ##width
h = 0.2 ##height
t = 0.015 ##thickness 
d = 0.03 ##diameter
r = 0.03 ##radius .....all dimensions in metres
E = 2*10**11  ##modulus of elasticity of structural steel as per ansys in Pa
𝛎 = 0.3 ## poison's ratio = 0.3
```

```python
𝛔_nom = 4*10**11 ##400MPa
K_t= 3 ##universal stress concentration constant
𝛔_max = K_t * 𝛔_nom
```

```python
A_gross = h * t
```

```python
A_gross
```

```python
A_net = (w-d)*t
A_net
```

```python
P = 𝛔_nom * A_gross
P
```

```python
𝛔_net = P/A_net
𝛔_net
```

```python
print ('Gross Area =' , A_gross, 'sq.m')
print ('Net Area =' , A_net, 'sq.m')
print ('Applied load =' , P, 'N')
print ('Nominal stress=' , 𝛔_nom, 'Pa')
print ('Stress concentration factor =' , K_t)
print ('Maximum stress=' , 𝛔_max, 'Pa')
print ('Net stress after cutting out a section of hole=' , 𝛔_net, 'Pa')
```

```python
##CIRCUMFERENTIAL STRESS
```

```python
θ = np.linspace(0, 2*(np.pi), 1000 )
𝛔_theta = 𝛔_nom*(1-2*np.cos(2*θ))
```

```python

plt.plot(θ, 𝛔_theta, color = 'green', linewidth = 0.6 )
plt.xlabel('Angle in degrees')
plt.ylabel('circumferential stress')
plt.title('Variation of circumferential stress around a circular hole in a 2D plate')
plt.grid(color ='black', linewidth = 0.5)
plt.show()
```

```python
plt.plot(np.degrees(θ), 𝛔_theta, color = 'green', linewidth = 0.6 )
plt.xlabel('Angle in degrees')
plt.ylabel('circumferential stress')
plt.title('Variation of circumferential stress around a circular hole in a 2D plate')
plt.grid(color ='black', linewidth = 0.5)
plt.show()
```
