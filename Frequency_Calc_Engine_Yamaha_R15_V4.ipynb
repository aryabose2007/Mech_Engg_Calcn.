```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MultipleLocator as Ml
```

  PRACTICAL DATA OF YAMAHA R 15 V4 ENGINE

MAX TORQUE is 14.2 Nm at 7500 RPM 
MAX POWER is 18.4 Ps at 10000 RPM 
Warm Idling	1,300 – 1,500 RPM	10.8 – 12.5 Hz (fires ~11 to 12 times per second)
Max Fuel Efficiency	4,000 – 5,000 RPM	33.3 – 41.6 Hz (fires ~33 to 42 times per second)
Peak Torque Output	7,500 RPM	62.5 Hz (fires 62.5 times per second)
Peak Power Output	10,000 RPM	83.3 Hz (fires 83.3 times per second)

```python
N_min = 500 ##minimum rpm
N_max = 10000 ## maximum rpm
```

```python
N = np.linspace(N_min, N_max, 3000)
```

```python
f_rot = N/60   ##rotational frequency
```

```python
f_firing = N/120  ##firing or combustion frequency
```

```python
plt.figure(figsize= (20,10), dpi= 100)
plt.plot (N, f_rot, color = 'magenta' , linewidth = 0.4)
plt.grid(color = 'black', linewidth = 0.4)
plt.gca().yaxis.set_major_locator(Ml(20))
plt.gca().xaxis.set_major_locator(Ml(500))
plt.xlabel('N = Range of RPM')
plt.ylabel('f = rotational frequency of engine (crankshaft)')
plt.show()
```

```python
plt.figure(figsize= (20,10), dpi= 100)
plt.plot (N, f_firing, color = 'magenta' , linewidth = 0.4)
plt.grid(color = 'black', linewidth = 0.4)
plt.gca().yaxis.set_major_locator(Ml(20))
plt.gca().xaxis.set_major_locator(Ml(500))
plt.ylabel('firing frequency of engine (crankshaft)')
plt.xlabel('N= range of RPM')
plt.savefig('Firing_Frequency_Engine.png')
plt.show()
```

```python
plt.figure(figsize= (20,10), dpi= 100)
plt.plot (N, 2*f_rot, color = 'magenta' , linewidth = 0.4)
plt.grid(color = 'black', linewidth = 0.4)
plt.gca().yaxis.set_major_locator(Ml(20))
plt.gca().xaxis.set_major_locator(Ml(500))
plt.ylabel('2nd order frequency of engine (connecting rod)')
plt.xlabel('N= range of RPM')
plt.savefig('2nd_order_Frequency_Engine.png')
plt.show()
```

```python
plt.figure(figsize= (30,30), dpi= 300)
plt.title('Ranges of frequencies for different RPM')
plt.plot (N, f_rot, color = 'magenta' , linewidth = 0.8, label = 'Rotation Frequency')
plt.plot (N, f_firing, color = 'blue' , linewidth = 0.8, label = 'Firing Frequency')
plt.plot (N, 2*f_rot, color = 'green' , linewidth = 0.8, label = '2nd Order Frequency')
plt.legend()
plt.xlabel('N= range of RPM')
plt.gca().yaxis.set_major_locator(Ml(20))
plt.gca().xaxis.set_major_locator(Ml(500))
plt.grid(color = 'black', linewidth = 0.8)
plt.savefig('Ranges of frequencies for different RPM.png')
plt.show()
```
